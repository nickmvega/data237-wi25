```python
import pandas as pd
import altair as alt
```

```python
ca = pd.read_csv("../data/ca-housing-umap.csv")
```

```python
ca.head()
```

```python
alt.data_transformers.disable_max_rows()
```

```python
brush = alt.selection_interval(resolve = "global")
```

```python
chart = alt.Chart(ca).mark_circle().encode(
    x='x:Q',
    y='y:Q',
    color=alt.condition(brush, 'ocean_proximity', alt.value('lightgray'))
).add_params(
    brush
).properties(
    width=350,
    height=400
)
```

```python
distribution1 = alt.Chart(ca).mark_bar().encode(
    x=alt.X('median_house_value:Q', bin=True, title="Median House Value"),
    y=alt.Y('count():Q', title=None),
    color=alt.value('lightgrey')
).add_params(
    brush 
).properties(
    width=300,
    height=150
)

distribution1_brush = alt.Chart(ca).mark_bar().encode(
    x=alt.X('median_house_value:Q', bin=True, title="Median House Value"),
    y=alt.Y('count():Q', title=None),
    color = "ocean_proximity"
).transform_filter(
    brush
).properties(
    width=300,
    height=150
)

distribution1 = alt.layer(
    distribution1,
    distribution1_brush
)
```

I decided to use Altair to build the dashboard given my familiarity and experiecne with Python and Altair. I find it both easier to render Altair charts through a Juptyer notebook and debug errors. While I choose Altair for this assignment, there are advantages for using a lower-level graphics tool like D3 over Altair. Two main advantages are that D3 is highly customizable to a user's preferences and D3 is able to handle larger datasets more efficently. The benefits of using Altair to add interactivity to data science workflows is that Altair is integrates extremely well with well-known python libraries such as pandas and numpy. Data integration is extremely important in data science workflows and Altair provides the resources to make data integration easy of any data source (csv, geojson, etc). An example of a situation were I would prefer to use D3 over Altair is when dealing with extremely large datasets that pandas and altair may have a hard time rendering. An example of a situation were I would prefer to use Altair over D3 would be when the dataset is smaller and easier to mainpulate. I feel that Altair and Pandas in this situation gives me full control over the data and being able to create an interactive visualization is much easier.
