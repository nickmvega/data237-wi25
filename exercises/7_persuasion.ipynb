```python
alt.Chart(round3).mark_line(
    point=True
).encode(
    x=alt.X('year:N', 
            axis=alt.Axis(
                values=[1975, 1984, 2000, 2006, 2024]  
            )),
    y=alt.Y('sales:Q')
).transform_filter(
    alt.FieldOneOfPredicate(
        field='year', 
        oneOf=[1975, 1984, 2000, 2006, 2024]
    )
).properties(
    width=600,
    height=400
)

```

```python
round4 = pd.read_csv("../data/persuasion-game/round4.csv")
```

```python
round4
```

```python
alt.data_transformers.disable_max_rows()
alt.Chart(round4).mark_bar().encode(
    x='Major:N',
    y=alt.Y('count():Q', axis=None),
    color='Admission:N',
    column='Gender:N'  
)
```

```python
base = alt.Chart(round4).transform_aggregate(
    count='count()',
    groupby=['Gender', 'Major', 'Admission']
).transform_joinaggregate(
    total='sum(count)',
    groupby=['Gender', 'Major']
).transform_calculate(
    percentage='100 * datum.count / datum.total'
)

# Create the visualization
base.mark_bar().encode(
    x='Major:N',
    y=alt.Y('percentage:Q', title='Percentage'),
    color='Admission:N',
    column='Gender:N'
)
```
